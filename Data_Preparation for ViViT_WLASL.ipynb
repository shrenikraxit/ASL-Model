```python
import os
import csv
from moviepy.editor import VideoFileClip

# Path to the folder containing the mp4 files
#folder_path = "/content/drive/MyDrive/WLASL dataset/videos"  # Replace with the actual folder path
folder_path = "/content/drive/MyDrive/WLASL_sample/videos_sample"  # Replace with the actual folder path

# Output CSV file path
#output_csv = "/content/drive/MyDrive/WLASL dataset/video_durations.csv"
output_csv = "/content/drive/MyDrive/WLASL_sample/video_durations.csv"

# Get a list of all mp4 files in the folder
video_files = [f for f in os.listdir(folder_path) if f.endswith('.mp4')]

# Open the CSV file for writing
with open(output_csv, mode='w', newline='') as file:
    writer = csv.writer(file)

    # Write the header row
    writer.writerow(["File Name", "Duration (seconds)","Frames"])

    # Loop through each video file and calculate its duration
    for video_file in video_files:
        video_path = os.path.join(folder_path, video_file)

        try:
            # Load the video file
            with VideoFileClip(video_path) as video:
                # Get the duration in seconds
                duration = video.duration
                fps = video.fps
                frames = fps*duration
                # Write the file name and duration to the CSV file
                writer.writerow([video_file, duration, frames])
                #print(f"Processed: {video_file} (Duration: {duration} seconds, FPS: {fps})")
        except Exception as e:
            # If an error occurs, print an error message and continue the loop
            writer.writerow([video_file, duration, 0])
            print(f"Error processing {video_file}: {str(e)}")




print(f"Durations saved to {output_csv}")
```

```python
import pandas as pd
import json

# Paths to the CSV and JSON files

video_duration_csv = "/content/drive/MyDrive/WLASL dataset/video_durations.csv"  # Replace with the correct path
wlasl_json = "/content/drive/MyDrive/WLASL dataset/WLASL_v0.3.json"  # Replace with the correct path
output_csv = "/content/drive/MyDrive/WLASL dataset/video_gloss_stat.csv"  # Replace with your desired output path
'''
video_duration_csv = "/content/drive/MyDrive/WLASL_sample/video_durations.csv"  # Replace with the correct path
wlasl_json = "/content/drive/MyDrive/WLASL_sample/WLASL_v0.3_sample.json"  # Replace with the correct path
output_csv = "/content/drive/MyDrive/WLASL_sample/video_gloss_stat.csv"  # Replace with your desired output path
'''

# Load the video duration CSV into a DataFrame
df_video = pd.read_csv(video_duration_csv)

# Load the wlasl.json data
with open(wlasl_json, 'r') as jsonfile:
    wlasl_data = json.load(jsonfile)

# Create an empty list to store the results
output_data = []

# Iterate over the gloss and instances in the JSON data
for entry in wlasl_data:
    gloss = entry['gloss']
    instances = entry['instances']

    for instance in instances:
        video_id = instance['video_id']  # Extract video_id
        file_name = f"{video_id}.mp4"  # Construct file name

        # Check if the video_id exists in the video_duration DataFrame
        matched_video = df_video[df_video['File Name'] == file_name]

        if not matched_video.empty:
            # For matched video, extract duration and frames
            file_name = matched_video['File Name'].values[0]
            duration = matched_video['Duration (seconds)'].values[0]
            frames = matched_video['Frames'].values[0]

            # Append the data to the output list
            output_data.append({
                'video_id': file_name,
                'duration': duration,
                'frames': frames,
                'gloss': gloss
            })

# Create a DataFrame from the output list
df_output = pd.DataFrame(output_data)

# Save the output DataFrame to a CSV file
df_output.to_csv(output_csv, sep='|', index=False)

print(f"Data successfully written to {output_csv}")
```
